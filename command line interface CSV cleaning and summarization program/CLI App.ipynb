```python
import pandas as pd
```

```python
df = pd.read_csv("API_NY.GDP.PCAP.CD_DS2_en_csv_v2_2.csv",skiprows=4)
df
```

```python
def clean_world_bank(df):
    year_cols = df.columns[4:]
    df[year_cols] = df[year_cols].apply(pd.to_numeric, errors="coerce")
    df = df.dropna(subset=year_cols,how='all')
    return df
```

```python
cleaned = clean_world_bank(df)
cleaned
```

```python
cleaned = cleaned.to_csv("cleaned_gdp_data.csv")
```
